```python
df_clean.coulumns=df_clean.columns.str.strip().str.lower().str.replace(" ","_")
```

```python
df_clean.columns
```

```python
text_columns=df_clean.select_dtypes(include="object").columns


for column in text_columns :
    df_clean[column]=df_clean[column].str.strip()

title_case_columns=["Gender","Marital_Status","Education","City","State","Region","Branch","Account_Type","Loan_Type"]


for column in title_case_columns:
        df_clean[column]=df_clean[column].str.title()
```

```python
df_clean[column]
```

```python
df_clean["Account_Type"]=df_clean["Account_Type"].str.strip().str.lower().replace({"saving":"Savings",
                                                                        "savings":"Savings",
                                                                        "current":"Current"})
```

```python
df_clean["Account_Type"].value_counts()
```

```python
df_clean.loc[~df_clean["Age"].between(18,100),"Age"]=np.nan
```

```python
df_clean["Age"]=df_clean["Age"].fillna(df_clean["Age"].median())
```

```python
df_clean.loc[~df_clean["Credit_Score"].between(300,900),"Credit_Score"]=np.nan
```

```python
df_clean["Credit_Score"]= df_clean["Credit_Score"].fillna(df_clean["Credit_Score"].median())
```

```python
avg_credit_score=df_clean['Credit_Score'].mean()
```

```python
df_clean["Education"]=df_clean["Education"].fillna(df_clean["Education"].mode()[0])
```

```python
df_clean['Occupation']=df_clean['Occupation'].fillna('NAN')
```

```python
df_clean["Annual_Income"]=df_clean["Annual_Income"].fillna(df_clean["Annual_Income"].median())
```

```python
df_clean["Balance"]=df_clean["Balance"].fillna(df_clean["Balance"].median())
```

```python
df_clean['Average_balance']=df_clean['Balance'].median()
```

```python
df_clean['Average_balance']
```

```python
df_clean["Loan_Type"]=df_clean["Loan_Type"].fillna("No loan")
```

```python
no_loan_mask=df_clean["Loan_Type"].eq("No loan")
df_clean.loc[no_loan_mask,"Loan_Amount"]=0
df_clean.loc[no_loan_mask,"emi"]=0
```

```python
loan_mask = ~no_loan_mask
df_clean.loc[loan_mask,"Loan_Amount"]=(df_clean.loc[loan_mask].groupby("Loan_Type")["Loan_Amount"].transform(lambda x:x.fillna(x.median())))
df_clean.loc[loan_mask,"emi"]=(df_clean.loc[loan_mask].groupby("Loan_Type")["emi"].transform(lambda x:x.fillna(x.median())))
```

```python
df_clean.loc[~df_clean["Customer_Satisfaction"].between(1,5),"Customer_Satisfaction"]=np.nan
df_clean["Customer_Satisfaction"]=(df_clean["Customer_Satisfaction"].fillna(df_clean["Customer_Satisfaction"].median()))
```

```python
df_clean['Average_customer_satisfaction']=df_clean['Customer_Satisfaction'].mean()
```

```python
df_clean["Age_Group"]=pd.cut(df_clean["Age"],bins=[17,25,35,45,55,65,100],labels=["18-25",
                                                                                 "26-35",
                                                                                 "36-45",
                                                                                 "46-55",
                                                                                 "56-65",
                                                                                 "65+"])
                                                                                 
```

```python

df_clean["Income_Group"]=pd.qcut(df_clean["Annual_Income"],q =4,labels=["Low Income",
                                                                        "Middle Income",
                                                                        "Upper Middle Income",
                                                                        "High Income"])
```

```python
df_clean["Balance_Category"]=pd.cut(df_clean["Balance"],bins=[-np.inf,100000,500000,1000000,np.inf],labels=["Low Balance",
                                                                                                            "Medium Balance",
                                                                                                            "High Balance",
                                                                                                            "Premium Balance"])
                                                                                                            
```

```python
reference_date = df_clean["Last_Transaction_Date"].max()


reference_date
```

```python
df_clean["Customer_Tenure_Years"]=(reference_date - df_clean["Customer_Since"]).dt.days/365.25
```

```python
recency_days = reference_date - df_clean['Last_Transaction_Date']
```

```python
df_clean['Active_Customer']=np.where((recency_days.dt.days <= 180)&(df_clean['Number_of_Transactions']>=10),"Yes","No")
```

```python
df_clean["Tenure_Category"]=pd.cut(df_clean["Customer_Tenure_Years"],bins=[-np.inf,1,3,5,10,np.inf],
                                   labels=["New Customer",
                                           "1-3 Years",
                                           "3-5 Years",
                                           "5-10 Years",
                                           "10+ Years"])
```

```python
df_clean["Transcations_Category"]=pd.cut(df_clean["Number_of_Transactions"],bins=[0,30,60,90,np.inf],
                                         labels=["Low",
                                                 "Moderate",
                                                 "High",
                                                 "Very High"])
```

```python
df_clean["Premium_Customer"]=np.where((df_clean["Balance"]>=1000000) & (df_clean["Annual_Income"] >= 2000000 ) & 
                                              (df_clean["Credit_Score"]>=750),"Yes","No")
```

```python
df_clean["Digital_Banking_User"]=np.where((df_clean["Internet_Banking"]=="Yes") | (df_clean["Mobile_App_User"] == "Yes"), 
                                              "Yes","No")
```

```python
df_clean["Net_Monthly_Cash_Flow"]=(df_clean["Monthly_Deposit"]-df_clean["Monthly_Withdrawal"])
```

```python
df_clean["Complaint_Category"]=pd.cut(df_clean["Complaints"],bins=[-1,0,2,5,np.inf],
                                      labels=["No Complaints",
                                              "Low","Medium","High"])
```

```python
churn_age = (df_clean.groupby('Age_Group',observed=False)['Churn'].apply(lambda x:(x =='Yes').mean()*100))
plt.figure(figsize=(10,6)) 
churn_age.plot(kind='bar')
plt.title("Churn Rate by Age_Group")
plt.xlabel('Age_Group')
plt.ylabel('Churn Rate(%)')
plt.show()
```

```python
churn_counts = df_clean["Churn"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    churn_counts,
    labels=churn_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.hist(
    df_clean["Age"],
    bins=20,
    edgecolor="black"
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="Account_Type",
    y="Balance"
)

plt.title("Account Balance by Account Type")
plt.xlabel("Account Type")
plt.ylabel("Balance")
plt.xticks(rotation=45)
plt.show()
```

```python
numeric_cols = [
    "Age",
    "Annual_Income",
    "Credit_Score",
    "Balance",
    "Monthly_Deposit",
    "Monthly_Withdrawal",
    "Number_of_Transactions",
    "Loan_Amount",
    "EMI",
    "Customer_Satisfaction"
]

corr = df_clean[numeric_cols].corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df_clean.sample(5000, random_state=42),
    x="Annual_Income",
    y="Balance",
    hue="Churn",
    alpha=0.6
)

plt.title("Annual Income vs Account Balance")
plt.xlabel("Annual Income")
plt.ylabel("Balance")
plt.show()
```
